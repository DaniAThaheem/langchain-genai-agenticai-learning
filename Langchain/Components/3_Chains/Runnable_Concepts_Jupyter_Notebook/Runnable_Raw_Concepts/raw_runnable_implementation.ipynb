### Standardizing the Components

```python
from abc import ABC, abstractmethod

class Runnable(ABC):
    def invoke():
        pass

```

```python
import random

class LLM(Runnable):
    def __init__(self):
        print("LLM created")

    def invoke(self, prompt):
        random_list = [
            "Erasmus Mundis is a prestigious scholorship requires time, efforts, intelligence, prayer and patience",
            "CSC, a chines scholarship where you have to find professors for the acceptence letters",
            "KAUST, a KSA based scholarship and there are two intakes, a spring and the other is fall"
        ]

        return {"response": random.choice(random_list)}
        

    def predict(self, prompt):
        random_list = [
            "Erasmus Mundis is a prestigious scholorship requires time, efforts, intelligence, prayer and patience",
            "CSC, a chines scholarship where you have to find professors for the acceptence letters",
            "KAUST, a KSA based scholarship and there are two intakes, a spring and the other is fall"
        ]

        return {"response": random.choice(random_list)}
```

```python
class PromptTemplate(Runnable):
    def __init__(self, template, input_variables):
        self.template = template
        self.input_variables = input_variables
    def invoke(self, input_variables):
        return self.template.format(**input_variables)
        
    def format(self, input_variables):
        return self.template.format(**input_variables)
        
```

### Creating a Class RunnableConnector for Creating Chain

```python
class RunnableConnector(Runnable):
    def __init__(self, runnable_list):
        self.runnable_list = runnable_list

    def invoke(self,input_data):
        for runnable in self.runnable_list:
            input_data = runnable.invoke(input_data)

        return input_data
        
```

```python
class StrOutputParser(Runnable):
    def invoke(self, input_data):
        return input_data["response"]
```

```python
template = PromptTemplate(
    template ="Give {sentences} sentence about this {scholarship} scholarship.",
    input_variables=["sentences", "scholarship"]
)
```

```python
llm = LLM()
```

```python
parser = StrOutputParser()
```

### Creating a Chain

```python
chain = RunnableConnector([template, llm, parser])
```

```python
result = chain.invoke({
    "sentences":"one",
    "scholarship":"CSC"
})
```

```python
result
```

### Creating Chain of Multiple Chains

```python
template1 = PromptTemplate(
    template ="Give {sentences} sentence about this {scholarship} scholarship.",
    input_variables=["sentences", "scholarship"]
)
```

```python
template2 = PromptTemplate(
    template ="Explanation of the scholarship {response}",
    input_variables=["response"]
)
```

```python
chain1 = RunnableConnector([template1, llm])
```

```python
chain2 = RunnableConnector([template2, llm, parser])
```

```python
chain = RunnableConnector([chain1, chain2])
```

```python
result = chain.invoke({
    "sentences":"one",
    "scholarship":"CSC"
})
```
