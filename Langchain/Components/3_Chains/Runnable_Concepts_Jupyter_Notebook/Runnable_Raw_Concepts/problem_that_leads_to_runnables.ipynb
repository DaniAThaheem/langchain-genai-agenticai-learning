### Creating a Class representing LLM Class in the Langchin

This is the LLM class where there is a predict method which is mocked to give the response to a prompt

```python
import random

class LLM:
    def __init__(self):
        print("LLM created")

    def predict(self, prompt):
        random_list = [
            "Erasmus Mundis is a prestigious scholorship requires time, efforts, intelligence, prayer and patience",
            "CSC, a chines scholarship where you have to find professors for the acceptence letters",
            "KAUST, a KSA based scholarship and there are two intakes, a spring and the other is fall"
        ]

        return {"response": random.choice(random_list)}
        
```

```python
llm = LLM()
```

```python
llm.predict("Erasmus")
```

```python
llm.predict("No I asked about Erasmus")["response"]
```

```python
llm.predict("No I asked about Erasmus")["response"]
```

```python
llm.predict("No I asked about Erasmus")["response"]
```

### Creating a Class representing PromptTemplate Class in the Langchain

This PromptTemplate class mock to give the prompt like the PromptTemplate class in the langchain

```python
class PromptTemplate:
    def __init__(self, template, input_variables):
        self.template = template
        self.input_variables = input_variables

    def format(self, input_variables):
        return self.template.format(**input_variables)
        
```

```python
template = PromptTemplate(
    template ="Give {sentences} sentence about this {scholarship} scholarship.",
    input_variables=["sentences", "scholarship"]
)
```

```python
prompt=template.format({
    "sentences":"one",
    "scholarship":"CSC"
})
```

### Creating a LLMChain Class 

The components are unstandardized. PromptTemplate Class has a format method and LLM class has a predict method. To make a chain of the Prompt and LLM we mock a built in class LLM to make a chain.

```python
class LLMChain:
    def __init__(self, llm, template):
        self.llm = llm,
        self.template = template

    def run(self, input_dict):
        prompt = self.template.format(input_dict)
        return llm.predict(prompt)
```

```python
llm = LLM()
```

```python
template = PromptTemplate(
    template ="Give {sentences} sentence about this {scholarship} scholarship.",
    input_variables=["sentences", "scholarship"]
)
```

```python
chain = LLMChain(llm, template)
```

```python
result = chain.run({
    "sentences":"one",
    "scholarship":"CSC"
})
```
